```python
import pandas as pd

df = pd.read_csv("final_food_delivery_dataset.csv")

# Filter only the required combinations
filtered = df[
    (
        (df["membership"] == "Gold") & (df["cuisine"] == "Indian")
    ) |
    (
        (df["membership"] == "Gold") & (df["cuisine"] == "Italian")
    ) |
    (
        (df["membership"] == "Regular") & (df["cuisine"] == "Indian")
    ) |
    (
        (df["membership"] == "Regular") & (df["cuisine"] == "Chinese")
    )
]

# Calculate total revenue for each combination
combo_revenue = (
    filtered
    .groupby(["membership", "cuisine"])["total_amount"]
    .sum()
    .reset_index(name="total_revenue")
)

# Sort to find highest revenue
combo_revenue = combo_revenue.sort_values("total_revenue", ascending=False)

print(combo_revenue)
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("final_food_delivery_dataset.csv")

# Convert order_date to datetime
df["order_date"] = pd.to_datetime(df["order_date"])

# Extract quarter
df["quarter"] = df["order_date"].dt.to_period("Q").astype(str)

# Calculate total revenue per quarter
quarter_revenue = (
    df.groupby("quarter")["total_amount"]
      .sum()
      .reset_index(name="total_revenue")
)

# Find quarter with highest revenue
top_quarter = quarter_revenue.sort_values(
    "total_revenue", ascending=False
).head(1)

print(quarter_revenue)
print("\nHighest revenue quarter:")
print(top_quarter)
```
